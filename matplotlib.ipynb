```python
import matplotlib.pyplot as plt
import numpy as np
```

Basic Line Plot

```python
x = np.array([1, 2, 3, 4])
y = np.array([10, 20, 30, 40])

plt.plot(x, y)
plt.title("Basic Line Plot")
plt.xlabel("X values")
plt.ylabel("Y values")
plt.show()
```

Line Styles and Colors

```python
x = np.array([1, 2,3, 4])
y = np.array([3, 8, 1, 10])

plt.plot(y, linestyle="dotted", color='green', marker='o')
plt.title("Styled Line Example")
plt.show()
```

#Multiple Lines in One Graph with Numpy and Matplotlib


```python
 x = np.array([1, 2, 3, 4])
 y1 = np.array([2, 4, 6, 8])
 y2 = np.array([1, 3, 5, 7])

 plt.plot(x, y1, label='line 1', color='red', marker='s')
 plt.plot(x, y2, label='line 2', color='blue', linestyle='--', marker='o')
 plt.title("Multiple Lines example")
 plt.xlabel("X Values")
 plt.ylabel("Y values")
 plt.legend()
 plt.show()
```

Bar Chart with Numpy and Matplotlib

```python
x = np.array(["A", 'B', 'C', 'D'])
y = np.array([5, 9, 1, 7])

plt.bar(x, y, color='orange')
plt.title("Bar Chart Example")
plt.xlabel("Categories")
plt.ylabel("Values")
plt.show()
```

#Scatter Plot with Numpy and Matplotlib

```python
x = np.array([1, 2, 3, 4, 5])
y = np.array([5, 4, 6, 3, 7])

plt.scatter(x, y, color='purple', s=100)
plt.title("Scatter Plot example")
plt.xlabel("X")
plt.ylabel("Y")
plt.show()
```

Histogram with Matplotlib and numpy

```python
data = np.array([3, 4, 5, 6, 7, 8, 9, 10])

plt.hist(data, bins=5, color='skyblue', edgecolor='black')
plt.title("Histogram Example")
plt.xlabel("X")
plt.ylabel("Y")
plt.show()
```

Sublpots(Multiple graphs together)


```python
x = np.array([1, 2, 3, 4])
y1 = np.array([2, 4, 6, 8])
y2 =np.array([3, 6, 9, 12])

plt.figure(figsize=(8,4)) #width and #height

plt.subplot(1,2,1)
plt.plot(x, y1, color='red')
plt.title('Plot 1')

plt.subplot(1, 2, 2)
plt.bar(x, y2, color='blue')
plt.title("Plot 2")

plt.tight_layout()
plt.show()
```
